```python
## Sum of array elements


class Solution:
    def sum(self, arr, n):
        count = 0

        for i in range(n):
            count += arr[i]

        return count
    


answer = Solution()
answer.sum([1,2,3,4,5], 5)
```

```python
## Count odd numbers in an array

class Solution:
    def countOdd(self, arr, n):
        # Your code goes here
        count = 0
        for i in range(n):
            if arr[i] % 2 != 0:
                count += 1

        return count


answer = Solution()
answer.countOdd([1,2,1,1,5,1], 6)
```

```python
## Check if the array is sorted



class Solution:
    def arraySortedOrNot(self, arr, n):
        present = arr[0]

        for i in range(1,n):
            if arr[i] < present:
                return False

            else:
                present = arr[i]

        return True




answer = Solution()
answer.arraySortedOrNot([5,4,6,7,8], 5)
```

```python
## Reverse an array


class Solution:
    def reverse(self, arr: list, n: int) -> None:
        def swap(arr,l,r):
            if l>=r:
                return

            temp = arr[l]
            arr[l] = arr[r]
            arr[r] = temp

            l+=1
            r-=1

            swap(arr,l,r)

        swap(arr,0,n-1)

        return arr


    ## l and r can be substituted by i which is l and n-i-1 which is r. mid point is going to be n/2 post which we dont swap





answer = Solution()
answer.reverse([1,2,3,4,5], 5)
```
