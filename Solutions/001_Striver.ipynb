```python
## Basic Hashing


def basic_hashing(arr, n):
    print(arr)
    hash_arr = {}
    for i in range(len(arr)):
        if arr[i] in hash_arr:
            hash_arr[arr[i]] += 1       
        else:
            hash_arr[arr[i]] = 1

    print(hash_arr[n])
            

arr = [1,2,1,3,2]
basic_hashing(arr, 1)
```

```python
def char_hash(arr, n):
    print(arr)
    hash = {}

    for char in arr:
        if char in hash:
            hash[char] += 1
        else:
            hash[char] = 1

    print(hash)


basic = "abcdabefc"
char_hash(basic, 'a')
```
