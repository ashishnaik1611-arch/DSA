```python
## Two Sum

class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        for i in range(len(nums)-1):
            check = 0
            check = target - nums[i]
            for j in range(i+1, len(nums)):
                if check == nums[j]:
                    return [i,j]

    def twoSum2(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i in range(len(nums)):
            check = target - nums[i]
            if check in seen:
                return [seen[check], i]

            seen[nums[i]] = i
            print(seen)


answer = Solution()
answer.twoSum2([2,7,11,15], 26)
```

```python
# Add Two Numbers - Local Driver

class ListNode:

    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


class Solution:
    def addTwoNumbers(self, l1: ListNode | None, l2: ListNode | None) -> ListNode | None:
        dummy = ListNode(0)
        current = dummy
        carry = 0

        while l1 or l2 or carry:
            num1 = l1.val if l1 else 0
            num2 = l2.val if l2 else 0

            total = num1+num2+carry

            digit = total%10
            carry = total//10

            current.next = ListNode(digit)
            current = current.next

            if l1:
                l1 = l1.next

            if l2:
                l2 = l2.next

        return dummy.next


# ==========================================
# Helper: Create Linked List
# ==========================================

def createLinkedList(values):

    dummy = ListNode(0)
    current = dummy

    for value in values:
        current.next = ListNode(value)
        current = current.next

    return dummy.next


# ==========================================
# Helper: Print Linked List
# ==========================================

def printLinkedList(head):

    current = head

    while current:

        print(current.val, end="")

        if current.next:
            print(" -> ", end="")

        current = current.next

    print()


# ==========================================
# Driver
# ==========================================

l1 = createLinkedList([2, 4, 3])
l2 = createLinkedList([5, 6, 4])

answer = Solution()

result = answer.addTwoNumbers(l1, l2)

printLinkedList(result)
```

```python
## Longest Substring without repearing characters


class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        seen = set()
        left = 0
        maximum = 0

        for right in range(len(s)):
            while s[right] in seen:
                seen.remove(s[left])
                left += 1

            seen.add(s[right])
            maximum = max(maximum, right-left+1)

        return maximum


answer = Solution()
answer.lengthOfLongestSubstring("abcabcbb")
```

```python
# Median of Two Sorted Arrays

class Solution:
    def findMedianSortedArrays(self, nums1: List[int], nums2: List[int]) -> float:
        sortedArray = []
        sortedArray = nums1 + nums2
        sortedArray.sort()

        n = len(sortedArray)
        mid = n // 2

        if n % 2 == 0:
            return (sortedArray[mid - 1] + sortedArray[mid]) / 2
        else:
            return sortedArray[mid]


answer = Solution()
answer.findMedianSortedArrays([1,3], [2])
```

```python
## 5. Longest Palindromic Substring


class Solution:
    def longestPalindrome(self, s: str) -> str:
        longest = ""

        if len(s) <= 1:
            longest = longest + s

        for i in range(len(s)):
            for j in range(i, len(s)):
                substring = s[i:j+1]
                if substring == substring[::-1]:
                    if len(longest) < len(substring):
                        longest = substring

        return longest


answer = Solution()
answer.longestPalindrome("babad")
```

```python
## 6. Zigzag Conversion

class Solution:
    def convert(self, s: str, numRows: int) -> str:
        if numRows == 1 or numRows >= len(s):
            return s
        
        goDown = True
        matrix = [[""] * len(s) for _ in range(numRows)]
        row = 0
        col = 0

        for char in s:
            matrix[row][col] = char

            if goDown:
                if row == numRows - 1:
                    goDown = False
                    row -= 1
                    col += 1

                else:
                    row += 1

            else:
                if row == 0:
                    goDown = True
                    row += 1

                else:
                    row -= 1
                    col += 2


        result = ""

        for r in range(numRows):
            for c in range(len(s)):

                if matrix[r][c] != "":
                    result = result + matrix[r][c]

        return result


        

answer = Solution()
answer.convert("ABC", 1)
```

```python
## 7. Reverse Integer


class Solution:
    def reverse(self, x: int) -> int:
        negative = 1
        if x < 0:
            negative = -1

        x = abs(x)

        reverse = 0
        result = 0

        while x > 0:
            digit = 0
            digit = x % 10
            reverse = (reverse * 10) + digit
            x = x //10
        
        if negative == -1:
            result = negative * reverse
            return result if result > -2**31 and result < 2**31 - 1 else 0

        else:
            return reverse if reverse > -2**31 and reverse < 2**31 -1 else 0




answer = Solution()
answer.reverse(1534236469)
```

```python
## 8. String to Integer (atoi)


class Solution:
    def myAtoi(self, s: str) -> int:
        s = s.strip()
        if not s:
            return 0

        negative = 1

        if s[0] == "-":
            negative = -1
            s = s[1:]
        elif s[0] == "+":
            s = s[1:]

        result = 0
        for char in s:
            if char.isdigit():
                result = result * 10 + int(char)
            else:
                break

        result = result * negative

        return result if result >= -2**31 and result <= 2**31 - 1 else (-2**31 if result < -2**31 else 2**31 - 1)

answer = Solution()
answer.myAtoi("-2147483648")
```

```python
## 9. Palindrome Number

class Solution:
    def isPalindrome(self, x: int) -> bool:
        check = str(x)

        if check == check[::-1]:
            return True
        else:
            return False


answer = Solution()
answer.isPalindrome(-10)
```

```python
## 10. Regular Expression Matching


class Solution:
    def isMatch(self, s: str, p: str) -> bool:
        if p in s:
            return True



answer = Solution()
answer.isMatch("aa", "a")
```
