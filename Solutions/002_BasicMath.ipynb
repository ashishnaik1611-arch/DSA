```python
## Count all digits of a number

class Solution:
    def countDigit(self, n):
        count = 0
        n = abs(n)

        while n > 0:
            n = n // 10
            count += 1

        print(count)

answer = Solution()
answer.countDigit(-1412)
```

```python
## Count number of odd digits in a number

class Solution:
    def countOddDigit(self, n):
        count = 0
        while n > 0:
            if n % 2 != 0:
                count += 1
            n = n //10

        return count


answer = Solution()
answer.countOddDigit(255132)
```

```python
## Reverse a number

class Solution:
    def reverseNumber(self, n):
        digit = 0
        reverse = 0

        while n > 0:
            digit = n % 10
            n = n // 10
            reverse = reverse * 10 + digit

        return reverse


answer = Solution()
answer.reverseNumber(25123)
```

```python
## Check Palindrome

class Solution:
    def isPalindrome(self, n, type):
        if type == 1:
            check = str(n)
            if check == check[::-1]:
                return True
            else:
                return False

        else:
            final = n
            digit = 0
            reverse = 0
    
            while n > 0:
                digit = n % 10
                n = n // 10
                reverse = reverse * 10 + digit
    
            if reverse == final:
                return True
            else:
                return False



answer = Solution()
answer.isPalindrome(121 , 0)
```

```python
## Return larget digit in a number

class Solution:
    def largestDigit(self, n):
        maximum = 0
        digit = 0
        while n > 0:
            digit = n % 10
            maximum = max(maximum, digit)
            n = n //10

        return maximum


answer = Solution()
answer.largestDigit(25)
```

```python
## Factorial of a given number

class Solution:
    def factorial(self, n):
        if n == 0:
            return 1

        total = n

        while n > 1:
            n = n-1
            total = total * n


        return total



answer = Solution()
answer.factorial(4)
```

```python
## Check if number is armstrong number


class Solution:
    def isArmstrong(self, n):
        total = 0
        check = n
        power = len(str(n))
        print(power)

        while n > 0:
            total = (n%10)**power + total
            n = n // 10

        print("check:", check, "total:", total)

        if total == check:
            return True
        else:
            return False


answer = Solution()
answer.isArmstrong(370)
```

```python
## Check for perfect numbers

class Solution:
    def isPerfect(self, n: int, type: str) -> bool:
        if type == "gpt":
            total = 0

            for i in range(1, n):
                if n % i == 0:
                    total += i

            return total == n

        else:
            if n <= 1:
                return False

            total = 1

            i = 2

            while i * i <= n:
                if n % i == 0:
                    total += i

                    if i != n // i:
                        total += n // i

                i += 1

            return total == n

            


answer = Solution()
answer.isPerfect(6, "gpt-1")
```

```python
## Check for prime numbers

class Solution:
    def isPrime(self, n):
        if n < 2:
            return False

        i = 2

        while i * i <= n:
            if n % i == 0:
                return False

            i += 1

        return True


answer = Solution()
print(answer.isPrime(5))
print(answer.isPrime(7))
print(answer.isPrime(10))
print(answer.isPrime(36))
```

```python
## Coount prime numbers up to n

class Solution:
    def primeUptoN(self, n):

        arr = [1]

        for num in range(2, n+1):
            if n % num == 0:
                arr.append(num)
            n+=1

        return len(arr)



answer = Solution()
answer.primeUptoN(10)
```

```python
## GCD of two numbers


class Solution:
    def GCD(self, n1, n2, type):

        if type == 1:

            gcd = 0

            for i in range(1, min(n1,n2)+1):
                if (n1%i == 0) and (n2%i == 0):
                    gcd = i

            return gcd

        elif type == 2:

            gcd = 0
            for i in range(min(n1,n2), 0, -1):
                if (n1%i == 0) and (n2%i == 0):
                    gcd = i
                    break

            return gcd

        elif type == 3:
            ## Euclidean formula

            gcd = 0
            cycle = max(n1,n2) - min(n1,n2)

            for i in range(1, cycle+1):
                if (n1%i == 0) and (n2%i == 0):
                    gcd = i

            return gcd

        else:
            ## Euclidean formula

            gcd = 0

            while n1>0 and n2>0:
                if n1>n2:
                    n1 = n1%n2

                else:
                    n2 = n2%n1

                if n1 == 0:
                    return n2
                else:
                    return n1

            



answer = Solution()
answer.GCD(20,40, 4)
```

```python
## LCM of two numbers


class Solution:
    def LCM(self, n1, n2):
        gcd = 0
        a = n1
        b = n2
        while n1 > 0 and n2 > 0:
            if n1>n2:
                n1 = n1%n2

            else:
                n2 = n2%n1

            if n1 == 0:
                gcd = n2

            else:
                gcd = n1

        lcm = a*b // gcd ## Formula for lcm is (n1*n2)//gcd

        return lcm
            




answer = Solution()
answer.LCM(4,6)
```
