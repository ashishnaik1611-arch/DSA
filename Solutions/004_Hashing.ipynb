```python
def sorting_bubble(nums):
    print(nums)
    n = len(nums)
    for i in range(len(nums)):
        swapped = False

        for j in range(n-i-1):
            if nums[j] > nums[j+1]:
                nums[j], nums[j+1] = nums[j+1],nums[j]
                swapped=True

        if not swapped:
            break

    return nums 




arr = [7,3,5,2,6,2,4,5,3]
sorting_bubble(arr)

```
