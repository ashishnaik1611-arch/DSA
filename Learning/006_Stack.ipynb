`Stack`

```python
### Basic

stack = []

# push
stack.append(10)
stack.append(20)
stack.append(30)

# pop
print(stack.pop())   # 30

# peek (top element)
print(stack[-1])     # 20

# check empty
print(len(stack) == 0)
```

```python
### With class

class Stack:
    def __init__(self):
        self.stack = []

    def push(self, val):
        self.stack.append(val)

    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        return "Stack is empty"

    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        return "Stack is empty"

    def is_empty(self):
        return len(self.stack) == 0

    def size(self):
        return len(self.stack)


# usage
s = Stack()
s.push(10)
s.push(20)
print(s.pop())   # 20
print(s.peek())  # 10
```

```python
### Using dequeue

from collections import deque

stack = deque()

stack.append(10)
stack.append(20)

print(stack.pop())   # 20
```

```python
### Valid parantheses -- https://leetcode.com/problems/valid-parentheses/?envType=problem-list-v2&envId=stack


class Solution:
    def isValid(self, s: str) -> bool:
        from collections import deque

        stack = deque()
        n = len(s)

        if n %2 != 0:
            return False

        for i in range(n):
            if s[i] in ('(', '{', '['):
                stack.append(s[i])
            else:
                if len(stack) > 0:
                    check = stack.pop()
                else:
                    return False
                if (s[i] == ')' and check == '(') or (s[i] == '}' and check == '{') or (s[i] == ']' and check == '['):
                    continue
                else:
                    return False

        if len(stack) != 0:
            return False
        
        return True
    

### GPTs response

class Solution:
    def isValid(self, s: str) -> bool:
        if len(s) % 2 != 0:
            return False

        stack = []
        mapping = {')': '(', '}': '{', ']': '['}

        for char in s:
            if char in mapping.values():  # opening
                stack.append(char)
            else:  # closing
                if not stack or stack.pop() != mapping[char]:
                    return False

        return not stack
```

```python
### Longest Valid paranthesis -- https://leetcode.com/problems/longest-valid-parentheses/?envType=problem-list-v2&envId=stack

class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack = [-1]
        total = 0

        for i in range(len(s)):
            if s[i] == '(':
                stack.append(i)
            
            else:
                stack.pop()

                if not stack:
                    stack.append(i)

                else:
                    length = i-stack[-1]
                    total = max(total, length)

        return total
                
```
