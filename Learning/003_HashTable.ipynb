```python
### Repeated DSA Sequence  -- https://leetcode.com/problems/repeated-dna-sequences/submissions/1999551885/?envType=problem-list-v2&envId=hash-function

class Solution:
    def findRepeatedDnaSequences(self, s: str) -> List[str]:
        n = len(s)
        if n < 10:
            return []

        seen = set()
        repeated = set()

        for i in range(n-9):
            sub = s[i:i+10]
            if sub in seen:
                repeated.add(sub)
            else:
                seen.add(sub)

        return list(repeated)
```

```python
class Solution:
    def shortestPalindrome(self, s: str) -> str:
        final = ""
        second_half = ""
        words = {}
        for ch in s:
            words[ch] = words.get(ch, 0) + 1
        
        print(words)

        for letter, count in words.items():
            if count % 2 == 0:
                final_num = count // 2
                final = final+(letter*final_num)

            else:
                final_num = (count // 2) + 1
                final = final+(letter*final_num)

        last_value = list(words.values())[-1]

        if last_value % 2 != 0:
            second_half = final[::-1]
            second_half = second_half[1:]
        else:
            second_half = final[::-1]

        final = final + second_half

        return final


answer = Solution()

s = "aacecaaa"

print(answer.shortestPalindrome(s))
```

```python
class Solution:
    def palindromePairs(self, words: List[str]) -> List[List[int]]:
        def isPalindrome(word, start, end):
            while start < end:
                if word[start] != word[end]:
                    
                    return False
                start += 1
                end -= 1
            return True

        n = len(words)
        output = []

        lookup = { w : i for i, w in enumerate(words)}

        for i in range(n):
            w = words[i]

            if w == "":
                for j in range(n):
                    if i!=j and isPalindrome(words[j], 0, len(words[j])-1):
                        output.append([i,j])
                        output.append([j,i])

                continue

            rev = w[::-1]

            if rev in lookup and lookup[rev] != i:
                output.append([i,lookup[rev]])
            
            for k in range(1, len(w)):
                if isPalindrome(w, 0, k-1) and w[k:][::-1] in lookup:
                    output.append([lookup[w[k:][::-1]], i])
                if isPalindrome(w, k, len(w)-1) and w[:k][::-1] in lookup:
                    output.append([i, lookup[w[:k][::-1]]])

        return output
                
            


    
class Solution_GPT:
    def palindromePairs(self, words):
        result = []

        word_map = {word: i for i, word in enumerate(words)}
        print(word_map)

        for i, word in enumerate(words):
            for j in range(len(word) + 1):

                left = word[:j]
                right = word[j:]

                if left == left[::-1]:

                    rev_right = right[::-1]

                    if rev_right in word_map and word_map[rev_right] != i:
                        result.append([word_map[rev_right], i])

                if j != len(word) and right == right[::-1]:

                    rev_left = left[::-1]

                    if rev_left in word_map and word_map[rev_left] != i:
                        result.append([i, word_map[rev_left]])

        return result


answer = Solution()
print(answer.palindromePairs(["abcd","dcba","lls","s","sssll"]))


# answer_gpt = Solution_GPT()
# print(answer_gpt.palindromePairs(["abcd","dcba","lls","s","sssll"]))
```
