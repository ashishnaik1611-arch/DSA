`Queue`

```python
### Using lists -- not recommended

queue = []

# enqueue
queue.append(10)
queue.append(20)

# dequeue
print(queue.pop(0))   # 10 ❌ slow (O(n))
```

```python
### Using dequeue

from collections import deque

queue = deque()

# enqueue
queue.append(10)
queue.append(20)

# dequeue
print(queue.popleft())   # 10 ✅ O(1)

# peek (front)
print(queue[0])

# check empty
print(len(queue) == 0)
```

```python
### Using Class

from collections import deque

class Queue:
    def __init__(self):
        self.q = deque()

    def enqueue(self, val):
        self.q.append(val)

    def dequeue(self):
        if not self.is_empty():
            return self.q.popleft()
        return "Queue is empty"

    def front(self):
        if not self.is_empty():
            return self.q[0]
        return "Queue is empty"

    def is_empty(self):
        return len(self.q) == 0

    def size(self):
        return len(self.q)


# usage
q = Queue()
q.enqueue(10)
q.enqueue(20)

print(q.dequeue())  # 10
print(q.front())    # 20
```
