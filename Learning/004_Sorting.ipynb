`Bubble Sort`

```python
def bubble_sort(nums):
    variable = True
    while variable:
        variable = False
        for i in range(len(nums) - 1):
            if nums[i] > nums[i+1]:
                nums[i], nums[i+1] = nums[i+1], nums[i]
                variable = True
            print("Iteration:", i, "--", nums)


l = [5,2,6,9,1,4,6,3]

bubble_sort(l)
print("Sorted list:", l)
```

`Insertion Sort`

```python
def insertion_sort(num):
    for i in range(1, len(num)):
        item = num[i]
        j = i-1
        print("Item", item)
        while j>=0 and num[j] > item:
            num[j+1] = num[j]
            j = j-1
            num[j+1] = item
            print(num)
        print(f"Iteration: {i}", num)

l = [5,2,6,9,1,4,6,3]

insertion_sort(l)
print("Sorted list:", l)
```

`Quick Sort`

```python
def partition(num, low, high):
    pivot = num[(low + high) // 2]
    i = low - 1
    j = high + 1

    while True:
        i+=1
        while num[i] < pivot:
            i+=1
        
        j-=1
        while num[j] > pivot:
            j-=1

        if i>=j:
            return j
        
        num[i], num[j] = num[j], num[i]


def quick_sort(num):
    def quick_recursion(item, low, high):
        if low < high:
            location = partition(item,low,high)
            quick_recursion(item, low, location)
            quick_recursion(item, location+1, high)

    quick_recursion(num, 0, len(num)-1)


l = [5,2,6,9,1,4,6,3]

quick_sort(l)
print("Sorted list:", l)
```

`Heap Sort`

```python
def max_heap(arr, n ,parent_idx):
    largest = parent_idx
    left = 2*parent_idx + 1
    right = 2* parent_idx + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != parent_idx:
        arr[largest], arr[parent_idx] = arr[parent_idx], arr[largest]
        max_heap(arr, n, largest)

def min_heap(arr, n, parent_idx):
    smallest = parent_idx
    left = 2 * parent_idx + 1
    right = 2 * parent_idx + 2

    if left < n and arr[left] < arr[smallest]:
        smallest = left

    if right < n and arr[right] < arr[smallest]:
        smallest = right

    if smallest != parent_idx:
        arr[smallest], arr[parent_idx] = arr[parent_idx], arr[smallest]
        min_heap(arr, n, smallest)

    
def heapsort(arr, maxHeap):
    if maxHeap:
        n = len(arr)
        for i in range(n//2 - 1, -1, -1):
            max_heap(arr, n, i)

        for i in range(n-1, 0, -1):
            arr[0], arr[i] = arr[i], arr[0]
            max_heap(arr, i, 0)

    else:
        n = len(arr)

        # Build min heap
        for i in range(n//2 - 1, -1, -1):
            min_heap(arr, n, i)

        # Extract elements
        for i in range(n-1, 0, -1):
            arr[0], arr[i] = arr[i], arr[0]   # move smallest to end
            min_heap(arr, i, 0)


l = [5,2,6,9,1,4,6,3]

maxHeap = False

heapsort(l, maxHeap)
print("Sorted list:", l)
```
