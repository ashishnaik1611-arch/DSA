```python
def fibonacci(n):
    if n == 0:
        return 0

    if n == 1:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)


n = 10

for i in range(n):
    print(fibonacci(i), end=" ")
```

```python
def isfrequent(s):
    freq = {}
    for c in s:
        freq[c] = freq.get(c,0)+1
    return freq

def isPalindrome(s):
    if s == s[::-1]:
        return True
    else:
        return False

print(isfrequent("hello world"))


print(isPalindrome("madaem"))
```
